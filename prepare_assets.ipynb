```python
# Asset Preparation
This notebook downloads the benchmark datasets (Set5, Set14, BSD100) and the pretrained
super-resolution models used in the project. It only needs to be run once; the files
are committed to the repository so that `main.ipynb` can run without external downloads.
```

```python
import tarfile
import urllib.request
from pathlib import Path

ROOT = Path.cwd()
DATA_DIR = ROOT / "data"
MODEL_DIR = ROOT / "models"
DATA_DIR.mkdir(exist_ok=True)
MODEL_DIR.mkdir(exist_ok=True)
print(ROOT)
```

```python
HF_URL = "https://huggingface.co/datasets/eugenesiow/{name}/resolve/main/data/{name}_HR.tar.gz"

for name in ["Set5", "Set14", "BSD100"]:
    target = DATA_DIR / name
    if target.exists():
        print(name, "already exists")
        continue
    archive = DATA_DIR / f"{name}_HR.tar.gz"
    urllib.request.urlretrieve(HF_URL.format(name=name), archive)
    with tarfile.open(archive) as tar:
        tar.extractall(DATA_DIR)
    (DATA_DIR / f"{name}_HR").rename(target)
    archive.unlink()
    print(name, len(list(target.iterdir())), "images")
```

```python
MODEL_SOURCES = {
    "EDSR": ("https://github.com/Saafke/EDSR_Tensorflow/raw/master/models", [2, 3, 4]),
    "ESPCN": ("https://github.com/fannymonori/TF-ESPCN/raw/master/export", [2, 3, 4]),
    "FSRCNN": ("https://github.com/Saafke/FSRCNN_Tensorflow/raw/master/models", [2, 3, 4]),
    "LapSRN": ("https://github.com/fannymonori/TF-LapSRN/raw/master/export", [2, 4]),
}

for model, (base_url, scales) in MODEL_SOURCES.items():
    for scale in scales:
        filename = f"{model}_x{scale}.pb"
        path = MODEL_DIR / filename
        if not path.exists():
            urllib.request.urlretrieve(f"{base_url}/{filename}", path)
        print(f"{filename:16s} {path.stat().st_size / 1e6:6.2f} MB")
```
